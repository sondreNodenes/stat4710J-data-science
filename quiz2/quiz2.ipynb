```python
import pandas as pd

elections = pd.read_csv('elections.csv')
elections
```

## Question 1 (5 pts)

Write a line of pandas code that returns the `elections` table sorted in ascending order by `"%"`. Store your result in a variable named `sorted_df`.

```python
# sort_values sorts the rows by the '%' column; ascending (smallest first) is the default.
sorted_df = elections.sort_values('%')
sorted_df
```

## Question 2 (2 pts)

Would calling `sorted_df.iloc[[0], :]` give the same result as `sorted_df.loc[[0], :]`? (Yes / No)

*Tip: run both and compare.*

**Answer: No**

```python
# No. iloc picks by POSITION, loc picks by index LABEL.
# After sorting, the first row (position 0) is David Cobb, 2004 (label 156),
# while the row with label 0 is Andrew Jackson, 1824, which is somewhere else in sorted_df.
display(sorted_df.iloc[[0], :])   # first row by position
display(sorted_df.loc[[0], :])    # row whose index label is 0
```

## Question 4 (5 pts)

Write one line of pandas code that returns a Series, where the index is the `"Party"` and the values are how many times that party won an election.

```python
# .loc[rows, column] picks the winning rows and the Party column in one step,
# then value_counts() counts how often each party appears.
elections.loc[elections["Result"] == "win", "Party"].value_counts()
```

## Question 5 (5 pts)

Write one line of pandas code that returns a Series whose index is the years and whose values are the number of candidates that participated in those years' elections.

```python
# groupby('Year') puts all candidates of the same year together; .size() counts the rows in each group.
elections.groupby('Year').size()
```

## Question 6 (5 pts)

Write a line of pandas code that creates a filtered DataFrame named `filtered_parties` from the `elections` dataset and keeps only the parties that have at least one election `%` more than 50%.

```python
# groupby("Party") makes one group per party. .filter(...) keeps a WHOLE group (all its rows)
# only if the function returns True for it. lambda group: ... gets one party's rows;
# (group["%"] > 50) is True/False per election, and .any() is True if at least one is True.
filtered_parties = elections.groupby("Party").filter(lambda group: (group["%"] > 50).any())
filtered_parties
```

## Question 7 (8 pts)

Write a line of pandas code that uses the `filtered_parties` DataFrame to return a new DataFrame with row indices that correspond to the year and columns that correspond to each party. Each entry should be the total percentage of votes for all the candidates that ran during that particular year for the specified party. Missing values (a party with no candidate that year) should be `0`.

Expected first rows:

| Year | Democratic | Democratic-Republican | National Union | Republican | Whig |
|---|---|---|---|---|---|
| 1824 | 0.000000 | 100 | 0.000000 | 0.000000 | 0.000000 |
| 1828 | 56.203927 | 0 | 0.000000 | 0.000000 | 0.000000 |

```python
# Same pivot_table pattern as in the lab (3d), but adding up '%' instead of counting:
#   index='Year'      -> one row per year
#   columns='Party'   -> one column per party
#   values='%'        -> the numbers to combine
#   aggfunc='sum'     -> add up the % of all of that party's candidates that year
#   fill_value=0      -> 0 instead of NaN when a party had no candidate that year
filtered_parties.pivot_table(index='Year', columns='Party', values='%', aggfunc='sum', fill_value=0)
```
